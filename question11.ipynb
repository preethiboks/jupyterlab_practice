```python
def compute_mean(values):
    mean = sum(values) / len(values)
    print(mean)

compute_mean([2,4,6,8])
```

The above `compute_mean` function calculates the mean (average) of a list of given numbers. It does so by summing all the numbers in the list and dividing the sum by the number of elements in the list. The function finally prints the computed mean. 


$$
\bar{x} = \frac{x_1 + x_2 + \cdots + x_n}{n} = \frac{1}{n} \sum_{i=1}^{n} x_i
$$
